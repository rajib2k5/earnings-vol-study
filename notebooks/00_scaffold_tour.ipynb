## 2. The envelope for the app
Python writes JSON, the React app only draws it. Every file has the same five keys. We pass a fixed time, so the output is the same on every run: 14:00 at +02:00 is 12:00 UTC.

```python
from datetime import datetime, timedelta, timezone

from earnings_vol import contract

when = datetime(2026, 10, 27, 14, 0, tzinfo=timezone(timedelta(hours=2)))
env = contract.envelope("demo", {"answer": 42}, generated_at=when,
                        source_commit="abc1234")
print(contract.dumps(env))
```
